```python
# imports
import pandas as pd
import snscrape.modules.twitter as sntwitter
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)

# Import the langdetect library
from langdetect import detect

# Import textblob library
from textblob import TextBlob

# maximum number of tweets to scrape
N = 500_000
```

```python
# A function to scrape n tweets containing desire word and retun a datafram of the tweets
def scrape(n, search_word):
    # Setting variables to be used below
    maxTweets = n

    # Creating list to append tweet data to
    tweets_list = []

    # Using TwitterSearchScraper to scrape data and append tweets to list
    for i,tweet in enumerate(sntwitter.TwitterSearchScraper(search_word).get_items()):
        if i > n:
            break
        tweets_list.append([tweet.date, tweet.id, tweet.content, tweet.user.username])
        
    # Creating a dataframe from the tweets list above
    df_temp = pd.DataFrame(tweets_list, columns=['Datetime', 'Tweet Id', 'Text', 'Username'])
    return df_temp
```

```python
# list of my EV makes and models

cars = [
'Tesla', 'Model 3', 'Model3', 'Model Y', 'Model X', 'Tesla Y', 'Tesla X',
'Mercedes-Benz AMG GLE63 4MATIC', 'EQB', 'EQS',
'BMW i3', 'BMW i4', 'i3', 'i4', 
'Ford Focus Electric', 'Focus Electric', 'MachE', 'Mach E', 'Mach-E', 'Ford Mach E', 
'Rivian', 'Chevrolet Bolt', 'Chevy Bolt EUV', 'Chevrolet Spark EV'
'Renault Fluence Z.E.', 'Renault Twizy', 'Renault Zoe', 
'Ioniq', 'Kia Soul EV', 'Kona', 'Niro', 'GV60', 'GV70', 'G80', 'G90', 
'Mazda MX-30', 'Mini Cooper EV', 'e Golf', 'egolf', 'e-Golf', 
'Volkswagen ID.4', 'Volkswagen ID 4', 'Volkswagen ID-4','Volkswagen ID4', 
'Toyota bZ4X', 'bZ4X', 'Subaru Solterra', 'Mitsubishi i-MiEV',
'Nissan Ariya', 'Nissan Leaf', 'Kia EV6', 'EV6', 'Honda Fit EV', 
'Jeep Wrangler 4xe', 'Jeep Wrangler', 'Jeep_EV',
'Audi Q4 E-Tron', 'e-Tron', 'q4', 'q8', 'SQ8',
'JAC J3 i EV'
]
```

```python
# Scraping N tweets for each car in the list 
df = pd.DataFrame()

for car in cars:
    df = df.append(scrape(N, car))
```

```python
df = df.reset_index(drop=True)
```

```python
df
```

```python
import langid
english_tweets = []
for tweet in df['Text']:
    # Language detection
    if langid.classify(tweet)[0] == 'en':
        english_tweets.append(tweet)
english_tweets = pd.DataFrame(english_tweets, columns=['tweets'])
```

```python
english_tweets
```

```python
# From mehranshakarami/AI_Spectrum

# precprcess tweet

for tweet in english_tweets['tweets']:
    tmpStr = tweet.split(' ')
    for i, word in enumerate(tmpStr):
        if word.startswith('@') and len(word) > 1:
            tmpStr[i] = '@user'
        elif word.startswith('http'):
            tmpStr[i] = "http"
    tweet = " ".join(tmpStr)
    english_tweets['tweets'][i] = tweet
    
```

```python
english_tweets
```

```python
# Keeping only tweets that have the desired words in their text. 
#(In the first place, tweets with the words in their username and text were scraped)

df_tweets = []
for i,tweet in enumerate(english_tweets['tweets']):
    for car in cars:
        if car in tweet:
            df_tweets.append(tweet)
df_tweets = pd.DataFrame(df_tweets)
```
